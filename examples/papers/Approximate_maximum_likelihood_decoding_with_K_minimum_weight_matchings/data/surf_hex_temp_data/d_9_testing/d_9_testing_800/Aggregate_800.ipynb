```julia
using JLD2
plist = Dict()
num_total_samples = 0
for file in readdir()
    if file[end-3:end] == "jld2"
        data = load(file)
        num_samples = data["num_samples"]
        num_total_samples += num_samples
        # println(num_samples)
        plist = merge(.+, plist, data["p0list"])
    end
end
println(num_total_samples)
```
