```python
import os
num = 4
path = f"d_9_testing_{num}"

dir_list = os.listdir(path)
for file in dir_list:
    file_split = file.split(".")
    if file_split[-1] == "jld2":
        file_split[-2] += f"_{num}"
        new_file_name = ".".join(file_split)
        os.rename(f"{path}/{file}", f"{path}/{new_file_name}")
```
