# d=7

```julia
f1 = "mwms_hex_7_7_0.596_0.6_1000320_200_3_v2.jld2"
f2 = "mwms_hex_7_7_0.601_0.605_1000320_200_3_v2.jld2"
f3 = "mwms_hex_7_7_0.606_0.606_1000320_200_3_v2.jld2"
f4 = "mwms_hex_7_7_0.607_0.607_1000032_200_3_v2.jld2";

```

```julia
d1 = load(f1)
```

```julia
d2 = load(f2)
```

```julia
d3 = load(f3)
```

```julia
d4 = load(f4)
```

```julia
timelist = merge(d1["timelist"], d2["timelist"], d3["timelist"], d4["timelist"])
```

```julia
p0list = merge(d1["p0list"], d2["p0list"], d3["p0list"], d4["p0list"])
sort(p0list)
```

```julia
f = "mwms_surf_hex_7_7_0.596_0.607_200_3_1000032.jld2"
num_samples = 1000032
drange = 7:2:7
σrange = 0.596:0.001:0.607
jldsave(f;
    num_samples = num_samples,
    p_mwms_list = p0list,
    t_mwms_list = timelist,
    σrange = σrange,
    drange = drange
)
```
